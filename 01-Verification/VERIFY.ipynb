```python
keywords = [
    "breast",
    "cancer",
    "tumor",
    "malignant",
    "survival",
    "mort",
    "diagnosis",
    "stage",
    "metast",
    "oncology"
]

datasets = {
    "Questionnaire": questionnaire,
    "Response": response,
    "Mortality": mortality,
    "Demographics": demographics
}

for name, df in datasets.items():
    print(f"\n{name}")
    found = []

    for col in df.columns:
        for key in keywords:
            if key in col.lower():
                found.append(col)
                break

    if found:
        print(found)
    else:
        print("No matching columns found.")
```

```python
# Search for variables beginning with common NHANES prefixes
prefixes = ["MCQ", "CNQ", "DIQ", "KIQ", "WHQ", "RHQ", "HUQ", "OCQ", "BPQ"]

for name, df in datasets.items():
    print(f"\n{name}")
    cols = [col for col in df.columns if any(col.startswith(p) for p in prefixes)]
    print(cols[:100])   # Show first 100 matching columns
    print(f"Total matching columns: {len(cols)}")
```
